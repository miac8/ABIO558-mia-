## 2) Compute relative expression levels for each treatment and gene
Use Pandas to do some standard qPCR calculations:
1. Average across technical replicates
2. Compute $\Delta$Ct by subtracting the gene Ct values from the endogenous control, Gapdh. This result should be a Pandas DataFrame - call it `delta_ct`
4. Compute $\Delta$$\Delta$Ct by subtracting the $\Delta$Ct values for the uninjected control from each sample's $\Delta$Ct value - call this DataFrame `dd_ct`
5. Compute the relative expression levels for experimental conditions vs. controls
* To compute relative expression from $\Delta \Delta$ Ct, use this formula:
* relative expression = 2^$\Delta \Delta$ Ct

```python
#define average dataframe, average across the technial replicates (multiple sample labels)
#compute delta_ct by subtracting the endogenous control (Gapdh)- row wise subtraction
#compute to delta-delta ct: subtract uninjected cotmrol from each sample look up pandas filter()
#then subtract control mean from each gene - column-wise 



tech_rep = ct_data.groupby('Sample').agg('mean')


#from beak_df.group avg mean tutorial

delta_ct = tech_rep.sub(tech_rep["Gapdh"], axis = 0)

#subtracts gapdh from gapdh-> row-wise subtraction rows = 0, 1, columns - axis

control_rows = delta_ct.filter(like='Uninjected', axis=0)
control_mean = control_rows.mean()

#control_mean = delta_ct.filter(like = "Uninjected", axis=0).mean()

#like= anything with the word uninjected will be subtracted only rows with uninjected 

dd_ct = delta_ct.sub(control_mean, axis=1)
#dd_ct = delta_ct.ct.sub(control_mean, axis=1)
#print(dd_ct)
relative_expression = 2 ** -dd_ct
print(relative_expression)
#rel_exp = 2**(dd_ct) 

#relative expression level double ** is to the power, so to the power of 2 (2^0, is 1) -> fold change 

#sample_type for each sample
```
